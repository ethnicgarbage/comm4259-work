### 15. Draw a histogram of profit in the year of 2000

```python
df1 = df.loc[df.Year == 2000]
```

```python
import seaborn as sns

sns.histplot(df1.Profit_millions.dropna(), kde = False).set_title("Histogram of Profit (in Millions) for the Year 2000")
```

### 16. Create a dataframe for Microsoft and Oracle. Draw a boxplot with the x axis showing two companies and y axis showing profit.

```python
df2 = df.loc[(df.Company == 'Microsoft') | (df.Company == 'Oracle')] 
```

```python
sns.boxplot(x = 'Company', y = 'Profit_millions', data = df2)
```

### 17. Draw two histograms side by side showing the profit for Microsoft and Oracle (FacetGrid)

```python
histProfitMO = sns.FacetGrid(df2, col = "Company")
histProfitMO = histProfitMO.map(sns.histplot, "Profit_millions")
```

### 18. Draw a line plot for Microsoft and Oracle
Create a lineplot to show Microsoft and Oracle's profit trend by year. Check the lineplot usage about x, y, and hue [here](https://seaborn.pydata.org/generated/seaborn.lineplot.html)

```python
sns.lineplot(x="Year", y="Profit_millions", hue="Company", marker="o", markersize = 10, data=df2)
```
